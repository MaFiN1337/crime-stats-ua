# Code to draw a graphic for Ukraine population


```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

dates = [
    datetime(2014, 1, 1),
    datetime(2015, 1, 1),
    datetime(2016, 1, 1),
    datetime(2017, 1, 1),
    datetime(2018, 1, 1),
    datetime(2019, 1, 1),
    datetime(2020, 1, 1),
    datetime(2021, 1, 1),
    datetime(2022, 1, 1),
    datetime(2022, 12, 1),
    datetime(2023, 12, 1),
    datetime(2024, 7, 1),
]

population = [44.2, 42.9,42.8,42.6,42.4,42.2,41.9,41.6,41.2,35,33.2,35]

plt.figure(figsize=(12, 6))
plt.plot(dates, population, color='red', linewidth=2, marker='o', markersize=5)

plt.xlabel('Рік', fontsize=12)
plt.ylabel('Млн осіб', fontsize=12)
plt.title('Чисельність населення України (2014-2024)', fontsize=14, fontweight='bold')

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
plt.gca().xaxis.set_major_locator(mdates.YearLocator())

plt.grid(True, alpha=0.3, linestyle='--')

plt.ylim(30, 47)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig('population_ukraine_2014_2024.png', dpi=300, bbox_inches='tight')
```
